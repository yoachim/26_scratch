```python
import matplotlib.pylab as plt
%matplotlib inline
import pandas as pd
import numpy as np
```

```python
# CPI from https://www.minneapolisfed.org/about-us/monetary-policy/inflation-calculator/consumer-price-index-1913-

dat = pd.read_csv("pay.txt", delimiter=",")
```

```python
pay_in_2025 = dat["pay"]*(dat["cpi"].values[-1]/dat["cpi"])

last_year = dat["year"].values[-1]
```

```python
fig, ax = plt.subplots()

ax.plot(dat["year"], dat["pay"], label="As Paid $")
ax.plot(dat["year"], pay_in_2025, label="CPI adjusted to %i" % last_year)
ax.set_xlabel("Year")
ax.set_ylabel("$")
lege = ax.legend()

ax.axhline(np.max(pay_in_2025), color="black", linestyle='--')
indx_max = np.where(pay_in_2025 == np.max(pay_in_2025))[0]
ax.plot(dat["year"].iloc[indx_max], pay_in_2025[indx_max], color="red", marker="*")
```
