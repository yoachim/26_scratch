```python
import numpy as np
import pandas as pd
from collections import Counter
```

```python
# read in dictionary. From github
words = pd.read_csv("words_alpha.txt")
words = words["a"].values
```

```python
# 
outlier_chars = 'pxihurdlgatsaencemfouwvik'
outlier_set = set(outlier_chars)
outlier_u = "".join(outlier_set)
oulier_u_len = len(outlier_u)
outlier_counter = Counter(outlier_chars)
```

```python
valid_word = []
for word in words:
    if type(word) != float:
        letter_counts = Counter(word)
        _temp = [outlier_counter[key] >= letter_counts[key] for key in letter_counts ]
        if all(_temp):
            valid_word.append(word)
```
