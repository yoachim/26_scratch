```python
import numpy as np
import pandas as pd
import glob
import matplotlib.pylab as plt
%matplotlib inline
```

```python
filenames = glob.glob("*.npz")
pop_names = [name.replace("comp_survey_v5.3.0_10yrs__", "").replace(".npz", "").replace("_obs", "") for name in filenames]
```

```python
pop_names
```

```python
summary = []
for pop_name in pop_names:
    mo = np.load("comp_survey_v5.3.0_10yrs__"+pop_name+"_obs.npz")
    mo_obs = mo["object_observations"].copy()
    mo.close()
    so_obs = pd.read_csv("sorcha_obs_" + pop_name + ".csv")
    summary.append({"pop": pop_name, "mo_n": len(mo_obs), "mo_u_id": np.size(np.unique(mo_obs["obj_id"])),
                   "so_n": len(so_obs), "so_u_id": np.size(np.unique(so_obs["ObjID"]))})
```

```python
pd.DataFrame(summary)
```
